```python
%matplotlib inline
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
sns.set(style="darkgrid")
```

```python
df = pd.read_csv('Vehiculos.csv', encoding='latin-1', sep=';')
```

```python
df.columns
```

```python
df = df.dropna()
```

```python
df['Relacion_Caballos_Consumo'] = df['Caballos'] / df['Consumo']
```

```python
df[['Modelo', 'Caballos', 'Consumo']].sort_values(by='Caballos', ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Obtener los 10 vehículos con más caballos
df_top = df[['Modelo', 'Caballos', 'Consumo']].sort_values(by='Caballos', ascending=False).head(10)

x = df_top['Consumo']
y = df_top['Caballos']

# Gráfico
fig, ax = plt.subplots(figsize=(10, 6))
scatter = ax.scatter(x, y, s=100, c=y, cmap='viridis')

ax.set_xlabel('Consumo')
ax.set_ylabel('Caballos')
ax.set_title('Relación Consumo vs Caballos (Top 10)')

plt.colorbar(scatter, label='Caballos')
plt.show()
```

```python
df['Precio en miles'] = pd.to_numeric(df['Precio en miles'].astype(str).str.replace(',', ''), errors='coerce')
```

```python
df['Relacion_Precio_Consumo'] = df['Precio en miles'] / df['Consumo']
```

```python
df[['Modelo', 'Precio en miles', 'Consumo']].sort_values(by='Precio en miles', ascending=False).head(10)
```

```python
import matplotlib.pyplot as plt
import numpy as np

df_top = df[['Modelo', 'Precio en miles', 'Consumo']].sort_values(by='Precio en miles', ascending=False).head(10)

x = df_top['Precio en miles']
y = df_top['Consumo']

# Gráfico
fig, ax = plt.subplots(figsize=(10, 6))
scatter = ax.scatter(x, y, s=100, c=y, cmap='viridis')

ax.set_xlabel('Precio en miles')
ax.set_ylabel('Consumo')
ax.set_title('Relación Precio vs Consumo  (Top 10)')

plt.colorbar(scatter, label='Precio en miles')
plt.show()
```

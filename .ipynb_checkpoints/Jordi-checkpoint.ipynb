```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
sns.set(style="darkgrid")
```

```python
#df = pd.read_csv('Vehiculos.csv', encoding='latin1')
df = pd.read_csv('Vehiculos.csv', encoding='latin1', sep=';')   #por defecto separa con , i nuestro csv separa con ;
```

```python
df.columns
```

```python
#convertir a numeric i canviar , per .df['Peso neto'] = pd.to_numeric(df['Peso neto'].astype(str).str.replace(',', ''), errors='coerce')
df['Caballos'] = pd.to_numeric(df['Caballos'].astype(str).str.replace(',', '.'), errors='coerce')

df['Relacio_Pes_Potencia'] = (df['Peso neto'] / df['Caballos']).round(2)
```

```python
#df['Modelo'].unique()  #esto muestra solo modelo
pd.set_option('display.max_rows', None)  #esto muestra siempre el maximo de linias
df[['Fabricante', 'Modelo', 'Peso neto', 'Caballos', 'Relacio_Pes_Potencia']].drop_duplicates()  #esto muestra fabricante, modelo 
```

```python
#ordenar de menor a major relacio i agafem els 10 primers
top10_rapidos = df.sort_values('Relacio_Pes_Potencia').head(10)

#Grafic de barres
plt.figure(figsize=(10, 5))
#sns.barplot(data=top10_rapidos, x='Relacio_Pes_Potencia', y='Model', palette='Blues_r')

sns.barplot(
    data=top10_rapidos, 
    x='Relacio_Pes_Potencia', 
    y='Modelo', 
    hue='Modelo', 
    palette='Blues_r', 
    legend=False
)

plt.title('Top 10 cotxes en Relacio Pes/Potencia')
plt.xlabel('Relacio Pes / Potencia (Menor = Mes potent/leuger)')
plt.ylabel('Model')
plt.show()
```
